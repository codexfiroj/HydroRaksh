### Verify Database Connectivity & Schema Inspection
We inspect the local SQLite database (`jalsuraksha_local.db`) to ensure observations and regional profiles exist.

```python
db_path = os.path.abspath('../../jalsuraksha_local.db')
assert os.path.exists(db_path), f'Database file not found at: {db_path}'

conn = sqlite3.connect(db_path)
tables = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table' ORDER BY name;", conn)
print('Connected successfully. Database tables:')
display(tables)
conn.close()
```
