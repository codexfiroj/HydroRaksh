# 01 — Official Data Source Audit & Open Repositories
**National Coverage, Licensing, and Ingestion Provenance**

Audits the official data sources powering JalSuraksha:
- **Central Ground Water Board (CGWB)**: Well water depth (mbgl) and GEC-2015 categorizations.
- **India-WRIS / NWIC**: Basin and telemetry observations.
- **India Meteorological Department (IMD)**: District daily/monthly rainfall.
- **Central Water Commission (CWC)**: Reservoir storage monitoring.
- **data.gov.in (OGD India)**: Agricultural and socio-demographic indicators.
- **Hugging Face & Kaggle**: Open research dataset fallbacks.

```python
import sys, os, yaml, pandas as pd
sys.path.insert(0, os.path.abspath('..'))
from src.data_loading import fetch_huggingface_dataset, fetch_kaggle_dataset

config_path = os.path.abspath('../configs/data_sources.yaml')
with open(config_path, 'r', encoding='utf-8') as f:
    sources_cfg = yaml.safe_load(f)

source_records = []
for key, info in sources_cfg.get('sources', {}).items():
    source_records.append({
        'Key': key,
        'Name': info.get('name'),
        'License': info.get('license'),
        'Refresh': info.get('refresh_frequency'),
        'Status': info.get('status'),
        'Auth Required': info.get('auth_required')
    })

sources_df = pd.DataFrame(source_records)
display(sources_df)
```
