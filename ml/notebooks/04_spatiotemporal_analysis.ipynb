# 04 — Spatiotemporal Groundwater Dynamics
**Seasonal Fluctuation, Monsoon Recharge Delta ($\Delta h$), and Spatial Autocorrelation**

Quantifies seasonal recharge recovery:
$$\Delta h = h_{\text{pre-monsoon}} - h_{\text{post-monsoon}}$$
A positive $\Delta h$ indicates water table recovery after monsoon precipitation.

```python
import sys, os, pandas as pd, numpy as np, matplotlib.pyplot as plt
sys.path.insert(0, os.path.abspath('..'))
from src.data_loading import load_groundwater_observations
from src.groundwater_analysis import calculate_seasonal_fluctuation

gw_df = load_groundwater_observations()
gw_df['recorded_date'] = pd.to_datetime(gw_df['recorded_date'])
gw_df['year'] = gw_df['recorded_date'].dt.year
gw_df['month'] = gw_df['recorded_date'].dt.month

# Aggregate pre-monsoon (April-June) and post-monsoon (October-December)
pre_monsoon = gw_df[gw_df['month'].isin([4, 5, 6])].groupby('district_id')['depth_to_water_mbgl'].mean()
post_monsoon = gw_df[gw_df['month'].isin([10, 11, 12])].groupby('district_id')['depth_to_water_mbgl'].mean()

recharge_df = pd.DataFrame({'pre_depth_mbgl': pre_monsoon, 'post_depth_mbgl': post_monsoon}).dropna()
recharge_df['water_table_rise_m'] = recharge_df['pre_depth_mbgl'] - recharge_df['post_depth_mbgl']
display(recharge_df.head(10))

print('Seasonal Water Table Rise Summary (meters):')
display(recharge_df['water_table_rise_m'].describe().to_frame().T)
```
