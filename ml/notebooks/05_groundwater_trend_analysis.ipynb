# 05 — Groundwater Monotonic Trend Analysis
**Mann-Kendall Non-Parametric Test & Sen's Slope Estimator**

Detects long-term monotonic water table depletion trends without assuming normal residuals:
$$S = \sum_{k=1}^{n-1} \sum_{j=k+1}^n \text{sgn}(x_j - x_k)$$
A statistically significant positive $S$ ($p < 0.05$) signifies that the depth mbgl is increasing (water table is depleting).

```python
import sys, os, pandas as pd, numpy as np
sys.path.insert(0, os.path.abspath('..'))
from src.data_loading import load_groundwater_observations
from src.groundwater_analysis import mann_kendall_trend_test, sens_slope_estimator

gw_df = load_groundwater_observations()
results = []
for district_id, grp in gw_df.groupby('district_id'):
    series = grp.sort_values('recorded_date')['depth_to_water_mbgl'].dropna().values
    if len(series) >= 4:
        mk = mann_kendall_trend_test(series)
        results.append({
            'district_id': district_id,
            'observations': mk['n'],
            's_stat': mk['s_stat'],
            'tau': mk['tau'],
            'p_value': mk['p_value'],
            'is_significant': mk['is_significant'],
            'trend': mk['trend']
        })

trends_df = pd.DataFrame(results)
display(trends_df)
```
