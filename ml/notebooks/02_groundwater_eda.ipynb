# 02 — Pan-India Groundwater Exploratory Data Analysis (EDA)
**Distribution, Depth Categories, and District Profiles**

Explores historical groundwater depth measurements ($meters\ below\ ground\ level\ -\ mbgl$) across monitoring wells and administrative units.

```python
import sys, os, pandas as pd, numpy as np, matplotlib.pyplot as plt
sys.path.insert(0, os.path.abspath('..'))
from src.data_loading import load_groundwater_observations, load_districts_hydrogeology

SAMPLE_MODE = False
gw_df = load_groundwater_observations(sample_mode=SAMPLE_MODE)
districts_df = load_districts_hydrogeology()

print(f'Loaded {len(gw_df)} observations across {gw_df["district_id"].nunique()} districts.')
display(gw_df.head(5))

print('Summary Statistics of Depth to Water (mbgl):')
display(gw_df['depth_to_water_mbgl'].describe().to_frame().T)
```

```python
plt.figure(figsize=(10, 4))
plt.hist(gw_df['depth_to_water_mbgl'].dropna(), bins=30, color='#0284c7', edgecolor='black', alpha=0.8)
plt.title('Distribution of Groundwater Depth to Water Level (mbgl)')
plt.xlabel('Depth below ground level (meters)')
plt.ylabel('Observation Count')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
