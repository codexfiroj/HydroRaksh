**District Extraction Stages and Regional Vulnerability Clusters**

Visualizes geographic patterns across Pan-India districts and identifies critical stress zones.

```python
import sys, os, pandas as pd, matplotlib.pyplot as plt
sys.path.insert(0, os.path.abspath('..'))
from src.data_loading import load_districts_hydrogeology

districts_df = load_districts_hydrogeology()
plt.figure(figsize=(10, 6))
scatter = plt.scatter(
    districts_df['longitude'],
    districts_df['latitude'],
    c=districts_df['stage_of_extraction_pct'].clip(upper=150),
    cmap='YlOrRd',
    s=districts_df['total_population'] / 50000,
    alpha=0.7,
    edgecolors='black',
    linewidth=0.5
)
plt.colorbar(scatter, label='Stage of Groundwater Extraction (%)')
plt.title('Pan-India Districts: Groundwater Stage of Extraction & Population Size')
plt.xlabel('Longitude (°E)')
plt.ylabel('Latitude (°N)')
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```
