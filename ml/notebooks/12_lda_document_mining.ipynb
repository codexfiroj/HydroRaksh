**Topic Extraction from CGWB Reports & Semantic Document Search**

Extracts 6 latent thematic topics across groundwater survey literature:
1. Deep Aquifer Depletion & Tubewell Densification
2. Industrial Effluent & Chemical Contamination
3. Canal Seepage Loss & Tail-End Deficits
4. Urban Impervious Runoff & Concretization
5. Geogenic Fluoride, Salinity & Arsenic Hazards

```python
import sys, os, pandas as pd
sys.path.insert(0, os.path.abspath('..'))
from src.document_mining import DocumentIntelligenceService

doc_service = DocumentIntelligenceService(num_topics=6)
doc_service.fit()

topics = doc_service.get_topics(top_n_words=6)
print('Extracted Hydrological Topics & Keywords:')
for t in topics:
    print(f"Topic {t['topic_id']}: {t['label']} -> {', '.join(t['top_words'])}")

# Semantic search query
query = 'fluoride contamination and salinity in hard rock aquifer'
results = doc_service.search_documents(query, top_k=3)
print(f'\nSemantic Search for: "{query}"')
display(pd.DataFrame(results)[['title', 'state', 'district', 'similarity_score']])
```
